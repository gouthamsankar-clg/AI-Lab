# Tic tac Toe Game

```python
print("Goutham Sankar - 1BM24CS105")

board = [[-1 for _ in range(3)] for _ in range(3)]
symbol = 'O'
full = 0

for r in board:
  print([cell if cell != -1 else ' ' for cell in r])

while (full <= 9):
    row = int(input("Enter row : "))
    col = int(input("Enter column : "))

    if board[row][col] == -1:
        board[row][col] = symbol
        full = full + 1

        for r in board:
            print([cell if cell != -1 else ' ' for cell in r])

        if board[row][0] == board[row][1] == board[row][2] and board[row][0] != -1:
            print(f"Player {symbol} won!")
            break

        if board[0][col] == board[1][col] == board[2][col] and board[0][col] != -1:
            print(f"Player {symbol} won!")
            break

        if (board[0][0] == board[1][1] == board[2][2] and board[0][0] != -1) or \
           (board[0][2] == board[1][1] == board[2][0] and board[0][2] != -1):
            print(f"Player {symbol} won!")
            break

        symbol = 'X' if symbol == 'O' else 'O'

    else:
        print("Invalid move")
```

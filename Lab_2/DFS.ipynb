```python
def read_board(name):
    print(f"Enter {name} state (3 rows, space-separated, 0 = blank):")
    b = [list(map(int, input().split())) for _ in range(3)]
    if sorted(sum(b, [])) != list(range(9)):
        raise ValueError("Board must contain digits 0-8 exactly once")
    return b

def solvable(a, b):
    def inv(board):
        t = [x for r in board for x in r if x]
        return sum(t[i] > t[j] for i in range(8) for j in range(i + 1, 8))
    return inv(a) % 2 == inv(b) % 2

def solve_dfs(start, goal):
    stack = [(start, 0)]
    visited = {str(start)}

    while stack:
        board, level = stack.pop()

        if board == goal:
            print("Goal found")
            print(f"Level: {level}")
            print("Final Board:")
            for row in board:
                print(row)
            print(f"Number of states visited: {len(visited)}")
            return True

        x, y = next((i, j) for i in range(3) for j in range(3) if board[i][j] == 0)

        for dx, dy in [(-1, 0), (1, 0), (0, -1), (0, 1)]:
            nx, ny = x + dx, y + dy
            if 0 <= nx < 3 and 0 <= ny < 3:
                nb = [r[:] for r in board]
                nb[x][y], nb[nx][ny] = nb[nx][ny], nb[x][y]
                key = str(nb)
                if key not in visited:
                    visited.add(key)
                    stack.append((nb, level + 1))

    print("No solution")
    print(f"Number of states visited: {len(visited)}")
    return False

start = read_board("start")
goal = read_board("goal")
if solvable(start, goal):
    solve_dfs(start, goal)
else:
    print("No solution (unsolvable parity)")
```

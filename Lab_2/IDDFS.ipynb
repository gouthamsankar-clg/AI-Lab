```python
import copy

def read_board(name):
    print(f"Enter {name} state (3 rows, space-separated, 0 = blank):")
    b = [list(map(int, input().split())) for _ in range(3)]
    if sorted(sum(b, [])) != list(range(9)):
        raise ValueError("Board must contain digits 0-8 exactly once")
    return b

def solvable(a, b):
    def inv(board):
        t = [x for r in board for x in r if x]
        return sum(t[i] > t[j] for i in range(8) for j in range(i + 1, 8))
    return inv(a) % 2 == inv(b) % 2

def dls(board, level, limit, visited, goal):
    global count
    count += 1

    if board == goal:
        print("Goal found")
        print(f"Level: {level}")
        print("Final Board:")
        for row in board:
            print(row)
        return True

    if level >= limit:
        return False

    key = str(board)
    if key in visited and visited[key] <= level:
        return False
    visited[key] = level

    x, y = next((i, j) for i in range(3) for j in range(3) if board[i][j] == 0)

    for dx, dy in [(-1, 0), (1, 0), (0, -1), (0, 1)]:
        nx, ny = x + dx, y + dy

        if 0 <= nx < 3 and 0 <= ny < 3:
            board[x][y], board[nx][ny] = board[nx][ny], board[x][y]

            if dls(board, level + 1, limit, visited, goal):
                return True

            board[x][y], board[nx][ny] = board[nx][ny], board[x][y]

    return False

def iddfs(start_board, goal):
    global count
    limit = 0
    while True:
        count = 0
        visited = {}
        print(f"\nTrying DLS with limit = {limit}")
        current_board_for_dls = copy.deepcopy(start_board)
        if dls(current_board_for_dls, 0, limit, visited, goal):
            print(f"Total states explored for solution: {count}")
            return True
        print(f"States explored at limit {limit}: {count}")
        limit += 1

count = 0
start = read_board("start")
goal = read_board("goal")

if solvable(start, goal):
    print("Running IDDFS...")
    iddfs(copy.deepcopy(start), goal)
else:
    print("No solution (unsolvable parity)")
```

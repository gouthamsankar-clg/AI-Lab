```python
import heapq

def get_h(state, goal):
    count = 0
    for i in range(3):
        for j in range(3):
            if state[i][j] != 0 and state[i][j] != goal[i][j]:
                count += 1
    return count

def find_blank(state):
    for i in range(3):
        for j in range(3):
            if state[i][j] == 0:
                return i, j

def get_neighbors(state):
    neighbors = []
    r, c = find_blank(state)
    moves = [("L", (0, -1)), ("U", (-1, 0)), ("R", (0, 1)), ("D", (1, 0))]

    for move_name, (dr, dc) in moves:
        nr, nc = r + dr, c + dc
        if 0 <= nr < 3 and 0 <= nc < 3:
            new_state = [list(row) for row in state]
            new_state[r][c], new_state[nr][nc] = new_state[nr][nc], new_state[r][c]
            neighbors.append((tuple(tuple(row) for row in new_state), move_name))
    return neighbors

def print_matrix(state):
    for row in state:
        print(" ".join(str(x) if x != 0 else "_" for x in row))

def a_star(initial, goal):
    pq = []
    h_init = get_h(initial, goal)
    heapq.heappush(pq, (0 + h_init, 0, initial, []))
    visited = set()

    while pq:
        f, g, current, path = heapq.heappop(pq)

        if current in visited:
            continue
        visited.add(current)

        print(f"Expanding Node at Depth g = {g}, h = {get_h(current, goal)}, f = {f}")
        print_matrix(current)
        print("-" * 15)

        if current == goal:
            print("Goal State Reached!")
            print(f"Total Moves: {len(path)}")
            print(f"Path taken: {' -> '.join(path)}")
            return

        for neighbor, move in get_neighbors(current):
            if neighbor not in visited:
                g_new = g + 1
                h_new = get_h(neighbor, goal)
                f_new = g_new + h_new
                heapq.heappush(pq, (f_new, g_new, neighbor, path + [move]))

if __name__ == "__main__":
    initial_state = ((2, 8, 3), (1, 6, 4), (7, 0, 5))
    goal_state = ((1, 2, 3), (8, 0, 4), (7, 6, 5))
    a_star(initial_state, goal_state)
```
